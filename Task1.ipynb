# Task 1: Iris Dataset Exploration

## Objective:
The goal of this task is to explore and visualize the Iris dataset to understand patterns and relationships between features.

## Dataset Loading

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("iris")
df.head()
```

## Data Exploration

```python
df.shape
df.columns
df.info()
df.describe()
```

## Data Visualization

```python
sns.scatterplot(x="sepal_length", y="sepal_width", hue="species", data=df)
plt.title("Sepal Length vs Width")
plt.show()
```

```python
df["petal_length"].hist()
plt.title("Petal Length Distribution")
plt.show()
```

```python
sns.boxplot(x="species", y="petal_length", data=df)
plt.title("Petal Length by Species")
plt.show()
```

## Final Insights

- Setosa has the smallest petal length  
- Virginica has the largest values  
- No missing data in dataset  
- Clear separation between species
